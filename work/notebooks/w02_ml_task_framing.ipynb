## 3. Success metric

*One metric you can defend. What number means 'good'?*

Metric: PR-AUC (Precision-Recall Area Under Curve) and F1-Score (targeted at the positive churn class).Defensible target: PR-AUC $> 0.65$ (or F1-Score $> 0.60$).Why: In imbalanced scenarios like churn detection, standard Accuracy is misleading (e.g., predicting all zeros gives high accuracy). PR-AUC ensures high precision (minimizing wasted intervention costs) while maintaining solid recall (catching true high-risk cases).

```python
from sklearn.metrics import classification_report, precision_recall_curve, auc

def evaluate_lane_performance(y_true, y_pred_probs, threshold=0.5):
    y_pred = (y_pred_probs >= threshold).astype(int)
    precision, recall, _ = precision_recall_curve(y_true, y_pred_probs)
    pr_auc = auc(recall, precision)
    print(f"PR-AUC: {pr_auc:.4f}")
    print("\nClassification Report:\n", classification_report(y_true, y_pred))
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Unit of Analysis: One row = One unique user/account observed over a specific monthly window.


```python
import pandas as pd
import numpy as np
data = {
    "user_id": [101, 102, 103, 104, 105],
    "tenure_months": [12, 2, 45, 1, 8],
    "monthly_charges": [65.5, 80.0, 20.0, 95.1, 40.0],
    "support_tickets_30d": [0, 4, 1, 5, 0],
    "is_churn": [0, 1, 0, 1, 0]
}

df = pd.DataFrame(data)
df.set_index("user_id", inplace=True)
print(f"Unit of Analysis: 1 row = 1 User ID")
df.head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Complexity: Customer churn behavior involves non-linear interactions across multiple features (e.g., high monthly charges combined with sudden spikes in support tickets and short tenure).

Limitations of If-Statements: Simple rules like if support_tickets > 3 then churn generate high false positive rates and fail to capture subtle, compounding behavioral signals across continuous dimensions (like declining activity frequency over time). Machine learning captures non-linear combinations and weighted probabilities dynamically.

```python

def rule_based_churn(row):
    return 1 if (row['support_tickets_30d'] > 3 and row['tenure_months'] < 6) else 0

df['rule_pred'] = df.apply(rule_based_churn, axis=1)
print("Rule-based predictions generated. ML models expand beyond this by capturing non-linear interactions across all feature dimensions.")
```
